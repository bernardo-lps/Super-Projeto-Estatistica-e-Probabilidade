```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import warnings
warnings.filterwarnings('ignore')

# Carregando dados
df = pd.read_csv('Sales Transaction v.4a.csv')
print("✓ Dados carregados:", df.shape)

# Preparar dados
dados = df[["ProductName", "Quantity", "Price", "Country"]].dropna().copy()
print("✓ Dados preparados:", dados.shape)

le_prod = LabelEncoder()
le_country = LabelEncoder()

dados["ProductName"] = le_prod.fit_transform(dados["ProductName"])
dados["Country"] = le_country.fit_transform(dados["Country"])

X = dados[["ProductName", "Quantity", "Price"]]
y = dados["Country"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
print("✓ Dados divididos - Train:", X_train.shape, "Test:", X_test.shape)

# Decision Tree
dt = DecisionTreeClassifier(random_state=42, max_depth=10)
dt.fit(X_train, y_train)
pred_dt = dt.predict(X_test)
print("\n✓ Decision Tree treinada!")

# KNN
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)
pred_knn = knn.predict(X_test)
print("✓ KNN treinada!")

# Resultados
print("\n" + "="*50)
print("RESULTADOS DOS MODELOS")
print("="*50)

for nome, pred in [("Decision Tree", pred_dt), ("KNN", pred_knn)]:
    print(f"\n📊 {nome}:")
    print(f"  Accuracy:  {accuracy_score(y_test, pred):.4f}")
    print(f"  Precision: {precision_score(y_test, pred, average='weighted', zero_division=0):.4f}")
    print(f"  Recall:    {recall_score(y_test, pred, average='weighted', zero_division=0):.4f}")
    print(f"  F1-Score:  {f1_score(y_test, pred, average='weighted', zero_division=0):.4f}")

print("\n✓ Análise concluída com sucesso!")
```
